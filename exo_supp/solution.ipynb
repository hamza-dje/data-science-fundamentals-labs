## *Importing libraries*

```python
import numpy as np
```

### **Generating data**

```python
products = ["Bread", "Milk", "Eggs", "Rice", "Oil"]
sales = np.array([
    [12, 5, 8, 20, 3],
    [15, 7, 10, 18, 4],
    [9, 6, 12, 25, 2],
    [20, 10, 9, 22, 6],
    [18, 8, 11, 19, 5],
    [25, 12, 15, 30, 8],
    [14, 6, 9, 21, 3],
])
```

### **Computing the total sales of each product over the week**

```python
sales_per_week = sales.sum(axis=0)
print(sales_per_week)
```

### **Computing the average of the daily total sales (all products combined)**

```python
daily_avg_sales = sales.mean(axis=1)
print(daily_avg_sales)
```

```python
total_daily_avg = sales.mean()
print(total_daily_avg)
```

### **Identifying the best-selling product of the week**

```python
print(f"Best selling product is {products[sales_per_week.argmax()]}")
```

### **Identifying the day with the highest total sales**

```python
days = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]

sales_per_day = sales.sum(axis=1)
print(sales_per_day)
```

```python
print(f"{days[sales_per_day.argmax()]} has the highest total sales")
```

### **Apply a 20% stock increase to prepare for next week**

```python
sales_copy = sales.copy()
sales_copy = np.ceil(sales_copy * 1.2)
print(sales_copy)
```

```python
print(sales_copy.sum(axis=0))
```
