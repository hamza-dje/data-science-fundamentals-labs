#### 2.1 Identifying which columns are numeric and which are array

```python
df_types = dict(df.dtypes)
df_types_final = {"numeric": [], "text": []}
for key, value in df_types.items():
    if str(value).startswith(("int", "float")):
        df_types_final["numeric"].append(key)
    else:
        df_types_final["text"].append(key)

df_types_final
```
