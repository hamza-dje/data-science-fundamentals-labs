## Exercise 3
### 1.

```python
grocery_prices = np.array([800, 1500, 450, 1200, 950])
hygiene_prices = np.array([650, 1200, 300, 980, 1450])

np.sort(hygiene_prices)[::-1]
```

### 2.

```python
np.append(hygiene_prices, 500)
np.delete(hygiene_prices, 0)
print(hygiene_prices)
```

### 3.

```python
store_prices = np.concatenate((grocery_prices, hygiene_prices))
print(f"Store prices: {store_prices}\nAverage price: {np.average(store_prices)}")
```
