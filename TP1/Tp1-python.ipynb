### 2. Functions

```python
def average_list(values):
    return sum(values) / len(values)

print(f"Average: {average_list([12, 15, 9, 18, 14])}")
```

## **Exercise 7**

```python
def max_grade(values) -> int | float:
    grades_copy = values
    grades_copy.sort()
    return grades_copy[-1]
```

## **Exercise 8**

```python
# is_even function implementation
def is_even(num) -> bool:
    return num % 2 == 0

# Using is_even on numbers between 1 and 30
for i in range(1, 31):
    print(f"{i} is {'even' if is_even(i) else 'odd'}")
```
